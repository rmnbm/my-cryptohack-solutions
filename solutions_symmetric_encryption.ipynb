```python
#
def shift_rows(s):
    s[0][1], s[1][1], s[2][1], s[3][1] = s[1][1], s[2][1], s[3][1], s[0][1]
    s[0][2], s[1][2], s[2][2], s[3][2] = s[2][2], s[3][2], s[0][2], s[1][2]
    s[0][3], s[1][3], s[2][3], s[3][3] = s[3][3], s[0][3], s[1][3], s[2][3]


def inv_shift_rows(s):
    s[0][1], s[1][1], s[2][1], s[3][1] = s[3][1], s[0][1], s[1][1], s[2][1]
    s[0][2], s[1][2], s[2][2], s[3][2] = s[2][2], s[3][2], s[0][2], s[1][2]
    s[0][3], s[1][3], s[2][3], s[3][3] = s[1][3], s[2][3], s[3][3], s[0][3]
    
    
# learned from http://cs.ucsb.edu/~koc/cs178/projects/JT/aes.c
xtime = lambda a: (((a << 1) ^ 0x1B) & 0xFF) if (a & 0x80) else (a << 1)


def mix_single_column(a):
    # see Sec 4.1.2 in The Design of Rijndael
    t = a[0] ^ a[1] ^ a[2] ^ a[3]
    u = a[0]
    a[0] ^= t ^ xtime(a[0] ^ a[1])
    a[1] ^= t ^ xtime(a[1] ^ a[2])
    a[2] ^= t ^ xtime(a[2] ^ a[3])
    a[3] ^= t ^ xtime(a[3] ^ u)


def mix_columns(s):
    for i in range(4):
        mix_single_column(s[i])


def inv_mix_columns(s):
    # see Sec 4.1.3 in The Design of Rijndael
    for i in range(4):
        u = xtime(xtime(s[i][0] ^ s[i][2]))
        v = xtime(xtime(s[i][1] ^ s[i][3]))
        s[i][0] ^= u
        s[i][1] ^= v
        s[i][2] ^= u
        s[i][3] ^= v

    mix_columns(s)


state = [
    [108, 106, 71, 86],
    [96, 62, 38, 72],
    [42, 184, 92, 209],
    [94, 79, 8, 54],
]


a = inv_mix_columns(state)
inv_shift_rows(state)
print(matrix2bytes(state))
```

```python
# Bringing It All Together
N_ROUNDS = 10

key        = b'\xc3,\\\xa6\xb5\x80^\x0c\xdb\x8d\xa5z*\xb6\xfe\\'
ciphertext = b'\xd1O\x14j\xa4+O\xb6\xa1\xc4\x08B)\x8f\x12\xdd'



def expand_key(master_key):
    """
    Expands and returns a list of key matrices for the given master_key.
    """

    # Round constants https://en.wikipedia.org/wiki/AES_key_schedule#Round_constants
    r_con = (
        0x00, 0x01, 0x02, 0x04, 0x08, 0x10, 0x20, 0x40,
        0x80, 0x1B, 0x36, 0x6C, 0xD8, 0xAB, 0x4D, 0x9A,
        0x2F, 0x5E, 0xBC, 0x63, 0xC6, 0x97, 0x35, 0x6A,
        0xD4, 0xB3, 0x7D, 0xFA, 0xEF, 0xC5, 0x91, 0x39,
    )

    # Initialize round keys with raw key material.
    key_columns = bytes2matrix(master_key)
    iteration_size = len(master_key) // 4

    # Each iteration has exactly as many columns as the key material.
    i = 1
    while len(key_columns) < (N_ROUNDS + 1) * 4:
        # Copy previous word.
        word = list(key_columns[-1])

        # Perform schedule_core once every "row".
        if len(key_columns) % iteration_size == 0:
            # Circular shift.
            word.append(word.pop(0))
            # Map to S-BOX.
            word = [s_box[b] for b in word]
            # XOR with first byte of R-CON, since the others bytes of R-CON are 0.
            word[0] ^= r_con[i]
            i += 1
        elif len(master_key) == 32 and len(key_columns) % iteration_size == 4:
            # Run word through S-box in the fourth iteration when using a
            # 256-bit key.
            word = [s_box[b] for b in word]

        # XOR with equivalent word from previous iteration.
        word = bytes(i^j for i, j in zip(word, key_columns[-iteration_size]))
        key_columns.append(word)

    # Group key words in 4x4 byte matrices.
    return [key_columns[4*i : 4*(i+1)] for i in range(len(key_columns) // 4)]


def decrypt(key, ciphertext):
    round_keys = expand_key(key) # Remember to start from the last round key and work backwards through them when decrypting

    # Convert ciphertext to state matrix
    state = bytes2matrix(ciphertext)
    

    # Initial add round key step
    add_round_key(state, round_keys[N_ROUNDS])

    for i in range(N_ROUNDS - 1, 0, -1):
        inv_shift_rows(state)
        sub_bytes(state, inv_s_box)
        add_round_key(state, round_keys[i])
        inv_mix_columns(state)
         # Do round

    # Run final round (skips the InvMixColumns step)
    inv_shift_rows(state)
    sub_bytes(state, inv_s_box)
    add_round_key(state, round_keys[0])

    # Convert state matrix to plaintext
    plaintext = matrix2bytes(state)

    return plaintext


print(decrypt(key, ciphertext))
```

**Lazy CBC**

In this CTF exercise, we immediately remark that the AES-CBC implementation's vulnerability is the Initialization Vector (IV) being equal to the Key. The goal is to use the fact that CBC mode XOR the IV and the first block in order to get the key. In order to do that the chosen-ciphertext attack uses a specific payload to get the *receive* function to return an error containing the decrypted payload. 

Payload structure: C1 || zero_block || C1
    
When decrypted by the server, it processes the 3 blocks:
- P'_1 = D(C1) XOR Key
- P'_2 = D(zero_block) XOR C1 (Generates junk bytes)
- P'_3 = D(C1) XOR zero_block = D(C1)

Because the server returns an error containing the full decrypted plaintext
(which fails ASCII decoding), we can extract P'_1 and P'_3. 
By XORing them together, D(C1) cancels out, leaving only the secret Key:
Key = P'_1 XOR P'_3.

```python
import requests
from Crypto.Util.strxor import strxor

# --- CONFIGURATION ---
BASE_URL = "https://aes.cryptohack.org/lazy_cbc"

def exploit_lazy_cbc():
    """
    Executes an error-based Chosen-Ciphertext Attack (CCA) against a vulnerable
    AES-CBC implementation where the Initialization Vector (IV) is equal to the Key.

    """
    print("[*] Starting the Lazy CBC attack...")

    # ==========================================
    # STEP 1: Obtain a valid ciphertext block (C1)
    # ==========================================
    # We encrypt 16 bytes (1 full block) to get a valid C1.
    plaintext_bytes = b"A" * 16
    
    response1 = requests.get(f"{BASE_URL}/encrypt/{plaintext_bytes.hex()}/").json()
    ciphertext_hex = response1["ciphertext"]
    
    # Isolate the first 32 hex characters (which represents exactly 16 bytes)
    C1_hex = ciphertext_hex[:32]
    print(f"[+] Retrieved Block C1: {C1_hex}")

    # ==========================================
    # STEP 2: Forge the malicious payload
    # ==========================================
    # Attack formula: C1 || 0 || C1
    zero_block_hex = (b"\x00" * 16).hex()
    payload_hex = C1_hex + zero_block_hex + C1_hex
    
    print(f"[*] Sending malicious payload (48 bytes): {payload_hex}")

    # ==========================================
    # STEP 3: Trigger the Oracle error
    # ==========================================
    # Send the payload to the receive/decrypt function
    response2 = requests.get(f"{BASE_URL}/receive/{payload_hex}/").json()
    
    if "error" not in response2:
        print("[-] Attack failed: No error returned from the server.")
        return

    print("[+] Oracle crashed successfully! Extracting memory leak...")

    # ==========================================
    # STEP 4: Extract and calculate the key
    # ==========================================
    # The error looks like: "Invalid plaintext: <hex_string>"
    error_message = response2["error"]
    leaked_plaintext_hex = error_message.split(" ")[-1]
    leaked_bytes = bytes.fromhex(leaked_plaintext_hex)

    # The leaked plaintext contains our 3 blocks. We isolate the 1st and 3rd.
    P1_prime = leaked_bytes[:16]
    P3_prime = leaked_bytes[32:48]

    # The mathematical magic: Key = P'_1 XOR P'_3
    key_bytes = strxor(P1_prime, P3_prime)
    key_hex = key_bytes.hex()
    
    print(f"[+] AES Key successfully extracted: {key_hex}")

    # ==========================================
    # STEP 5: Retrieve the Flag
    # ==========================================
    response3 = requests.get(f"{BASE_URL}/get_flag/{key_hex}/").json()
    
    if "plaintext" in response3:
        flag_hex = response3["plaintext"]
        flag = bytes.fromhex(flag_hex).decode('utf-8')
        print(f"\n[🏆] FLAG FOUND: {flag}\n")
    else:
        print("[-] Error while retrieving the flag:", response3)


# Entry point of the script
if __name__ == "__main__":
    exploit_lazy_cbc()
```
