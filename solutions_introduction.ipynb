```python
#XOR Properties
KEY1_hex = "a6c8b6733c9b22de7bc0253266a3867df55acde8635e19c73313"
KEY2_KEY1_hex = "37dcb292030faa90d07eec17e3b1c6d8daf94c35d4c9191a5e1e"
KEY2_KEY3_hex = "c1545756687e7573db23aa1c3452a098b71a7fbf0fddddde5fc1"
FLAG_KEYS_hex = "04ee9855208a2cd59091d04767ae47963170d1660df7f56f5faf"

KEY2 = xor(bytes.fromhex(KEY1_hex), bytes.fromhex(KEY2_KEY1_hex))
KEY3 = xor(KEY2, bytes.fromhex(KEY2_KEY3_hex))
KEY123 = xor(bytes.fromhex(KEY1_hex), KEY2, KEY3)
FLAG = xor(bytes.fromhex(FLAG_KEYS_hex), KEY123)
print(FLAG.decode())
```

```python
#Favourite Byte
hex_string = "73626960647f6b206821204f21254f7d694f7624662065622127234f726927756d"
ciphertext = bytes.fromhex(hex_string)
for i in range(256):
    flag = xor(ciphertext, i)
    if b"crypto{" in flag:
        print(flag.decode(), end ="")
```
